## 2. Variables and Data Types

**Naming rules**
- Must start with a letter or underscore (`_`)

- Cannot be a keyword (`if`, `for`, `class`, ...)

## 3. Integers, Floats, Strings and Booleans

| Type | Meaning | Example |
|------|---------|---------|
| `int` | Whole numbers | `10`, `-3`, `0` |
| `float` | Decimal numbers | `3.14`, `-0.5` |
| `str` | Text | `"hello"`, `'hi'` |
| `bool` | `True` / `False` | `True` |

Use `type()` to check the type of a value.

```python
# Integers
count = 42
big = 1_000_000          # underscores for readability
print(count, big, type(count))

# Floats
pi = 3.14159
sci = 15.5e2              # 1500.0
print(pi, sci, type(pi))

# Strings
s1 = "Hello"
s2 = 'World'
s3 = '''This is a
multi-line string'''
print(s1, s2)
print(s3)
print(type(s1))

# Booleans
t = True
f = False
print(t, f, type(t))
print(5 > 3, 5 == 3)
```

### Truthy and Falsy values
Every value has a boolean meaning. These are **falsy**: `0`, `0.0`, `""`, `None`, `[]`, `{}`. Everything else is **truthy**.

```python
for value in [0, 1, -1, 0.0, "", "hi", None, [], [0]]:
    print(f"{value!r:>6} -> {bool(value)}")
```

### Floating-point precision
Computers store floats in binary, so some decimals are approximations.

```python
print(0.1 + 0.25)               # 0.30000000000000004
print(round(0.1 + 0.25, 1))     # 0.3
print(type(10 / 2))            # / always gives float
print(10 / 2)            # / always gives float
print(type(10 // 2))           # // gives int for int inputs
print(10 // 2)           # / always gives float          # // gives int for int inputs
```

## 4. Type Conversion

Changing a value from one type to another.

- **Implicit:** Python does it automatically (e.g. `int + float -> float`)
- **Explicit (casting):** you do it with `int()`, `float()`, `str()`, `bool()`

```python
# Implicit conversion
result = 5 + 2.5
print(result, type(result))

# Explicit conversion
print(int("123") + 1)        # str -> int
print(float("3.5") * 2)      # str -> float
print(str(100) + " marks")   # int -> str
print(int(7.9))              # float -> int (truncates, does NOT round)
print(float(5))              # int -> float
print(bool(0), bool("0"))    # False True
print(int(True), int(False)) # 1 0
```

```python
# Common error: int("abc") fails. Handle it safely with try/except.
for text in ["42", "3.14", "abc", ""]:
    try:
        print(f"{text!r} -> {int(text)}")
    except ValueError:
        print(f"{text!r} -> cannot convert to int")
```

## 5. Operators

### 5.1 Arithmetic
`+  -  *  /  //  %  **`

```python
a, b = 17, 5
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)     # true division
print("a // b =", a // b)    # floor division
print("a % b  =", a % b)     # remainder
print("a ** b =", a ** b)    # power
```

### 5.2 Comparison (return `True`/`False`)
`==  !=  >  <  >=  <=`

```python
x, y = 10, 20
print(x == y, x != y, x > y, x < y, x >= 10, y <= 15)
```

### 5.3 Logical
`and`, `or`, `not`

```python
age = 25
has_id = True
print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
```

### 5.4 Assignment
`=  +=  -=  *=  /=  //=  %=  **=`

```python
n = 10
n += 5;  print(n)
n -= 3;  print(n)
n *= 2;  print(n)
n /= 4;  print(n)
n **= 2; print(n)
```

### 5.5 Membership and Identity
`in`, `not in`, `is`, `is not`

```python
print("py" in "python")
print("z" not in "python")

p = [1, 2]
q = [1, 2]
r = p
print(p == q)   # same value
print(p is q)   # different objects
print(p is r)   # same object
```

### 5.6 Operator precedence
Order (high to low): `()` > `**` > `*  /  //  %` > `+  -` > comparisons > `not` > `and` > `or`.
Use parentheses to make intent clear.

```python
print(2 + 3 * 4)        # 14
print((2 + 3) * 4)      # 20
print(2 ** 3 ** 2)      # 512 (right to left)
print(-3 ** 2)          # -9
```

# 🛠️ Practical 1: Basic Calculator

We will build it in two steps:
1. A `calculate()` function that does the math
2. An interactive loop that talks to the user

```python
def calculate(a,op,b):
    """Return the result of a <op> b, or an error message."""
    if op == "+":
        return a + b
    elif op == "-":
        return a - b
    elif op == "*":
        return a * b
    elif op == "/":
        return "Error: division by zero" if b == 0 else a / b
    elif op == "//":
        return "Error: division by zero" if b == 0 else a // b
    elif op == "%":
        return "Error: division by zero" if b == 0 else a % b
    elif op == "**":
        return a ** b
    else:
        return f"Error: unknown operator '{op}'"

# Quick tests
tests = [(10, "+", 5), (10, "-", 5), (10, "*", 5), (10, "/", 4),
         (10, "//", 4), (10, "%", 4), (2, "**", 8), (5, "/", 0), (5, "&", 2)]
for a, op, b in tests:
    print(f"{a} {op} {b} = {calculate(a, op, b)}")
```

```python
RUN_INTERACTIVE = True
```

```python
def calculator_app():
    print("=" * 30)
    print("   BASIC CALCULATOR")
    print("   Operators: + - * / // % **")
    print("=" * 30)
    while True:
        try:
            a = float(input("First number : "))
            op = input("Operator     : ").strip()
            b = float(input("Second number: "))
        except ValueError:
            print("Please enter valid numbers.\n")
            continue

        result = calculate(a, op, b)
        print(f"Result: {result}\n")

        again = input("Calculate again? (y/n): ").strip().lower()
        if again != "y":
            print("Goodbye!")
            break

if RUN_INTERACTIVE:
    calculator_app()
else:
    print("Set RUN_INTERACTIVE = True (first cell) and re-run this cell to use the calculator.")
```

# 🛠️ Practical 2: Programs for String Operations

```python
# Program 1: Reverse a string and check palindrome
def is_palindrome(s):
    cleaned = "".join(ch.lower() for ch in s if ch.isalnum())
    return cleaned == cleaned[::-1]

for w in ["madam", "Python", "A man, a plan, a canal: Panama"]:
    print(f"{w!r:40} palindrome? {is_palindrome(w)}")
```

```python
# Program 2: Count vowels and consonants
def count_vowels_consonants(s):
    vowels = consonants = 0
    for ch in s.lower():
        if ch.isalpha():
            if ch in "aeiou":
                vowels += 1
            else:
                consonants += 1
    return vowels, consonants

v, c = count_vowels_consonants("Python Programming")
print(f"Vowels: {v}, Consonants: {c}")
```

```python
# Program 3: Character frequency
def char_frequency(s):
    freq = {}
    for ch in s.lower():
        if ch != " ":
            freq[ch] = freq.get(ch, 0) + 1
    return freq

freq = char_frequency("hello world")
for ch, n in sorted(freq.items(), key=lambda kv: (-kv[1], kv[0])):
    print(f"{ch}:({n})")
```

```python
# Program 4: Word operations
sentence = "python is simple and python is powerful"
words = sentence.split()
print("Word count   :", len(words))
print("Unique words :", sorted(set(words)))
print("Longest word :", max(words, key=len))
print("Title case   :", sentence.title())
print("Reverse words:", " ".join(reversed(words)))
print("Reverse each :", " ".join(w[::-1] for w in words))
```

```python
# Program 5: Caesar cipher
def caesar(text, shift):
    result = ""
    for ch in text:
        if ch.isalpha():
            base = ord("A") if ch.isupper() else ord("a")
            result += chr((ord(ch) - base + shift) % 26 + base)
        else:
            result += ch
    return result

secret = caesar("Hello, World!", 3)
print(secret)
print(caesar(secret, -3))
```

```python
# Program 6: Remove duplicates while keeping order + toggle case
def remove_duplicates(s):
    seen = ""
    for ch in s:
        if ch not in seen:
            seen += ch
    return seen

print(remove_duplicates("programming"))
print("Hello World".swapcase())
```

# 🛠️ Practical 3: Type Conversion Practice

```python
# Exercise table: value -> different conversions
samples = ["10", "3.99", "True", "0", "", " 42 "]

print(f"{'value':>8} | {'int()':>8} | {'float()':>8} | {'bool()':>6}")
print("-" * 40)
for v in samples:
    try:
        i = int(v)
    except ValueError:
        i = "error"
    try:
        f = float(v)
    except ValueError:
        f = "error"
    print(f"{v!r:>8} | {str(i):>8} | {str(f):>8} | {str(bool(v)):>6}")
```

```python
# Safe converter function
def to_number(text):
    """Convert text to int if possible, else float, else None."""
    text = text.strip()
    try:
        return int(text)
    except ValueError:
        try:
            return float(text)
        except ValueError:
            return None

for t in ["15", "15.5", "  7 ", "1e3", "hello"]:
    val = to_number(t)
    print(f"{t!r:10} -> {val!r:8} {type(val).__name__}")
```

## App 1: BMI Calculator

```python
RUN_INTERACTIVE=True
```

```python
def bmi_calculator(weight_kg, height_m):
    bmi = weight_kg / (height_m ** 2)
    if bmi < 18.5:
        category = "Underweight"
    elif bmi < 25:
        category = "Normal"
    elif bmi < 30:
        category = "Overweight"
    else:
        category = "Obese"
    return bmi, category

if RUN_INTERACTIVE:
    w = float(input("Weight (kg): "))
    h = float(input("Height (m): "))
else:
    w, h = 68, 1.72

bmi, cat = bmi_calculator(w, h)
print(f"BMI = {bmi:.2f} -> {cat}")
```

## App 2: Temperature Converter

```python
def convert_temperature(value, unit):
    unit = unit.upper()
    if unit == "C":
        return f"{value}°C = {value * 9/5 + 32:.2f}°F = {value + 273.15:.2f}K"
    elif unit == "F":
        c = (value - 32) * 5/9
        return f"{value}°F = {c:.2f}°C = {c + 273.15:.2f}K"
    elif unit == "K":
        c = value - 273.15
        return f"{value}K = {c:.2f}°C = {c * 9/5 + 32:.2f}°F"
    return "Unknown unit. Use C, F or K."

print(convert_temperature(100, "C"))
print(convert_temperature(98.6, "F"))
print(convert_temperature(300, "K"))
```

## App 3: Bill Splitter with Tip

```python
def split_bill(amount, tip_percent, people):
    tip = amount * tip_percent / 100
    total = amount + tip
    per_person = total / people
    print("=" * 30)
    print(f"{'Bill amount':<15}: {amount:>10,.2f}")
    print(f"{'Tip (' + str(tip_percent) + '%)':<15}: {tip:>10,.2f}")
    print(f"{'Total':<15}: {total:>10,.2f}")
    print(f"{'Each of ' + str(people):<15}: {per_person:>10,.2f}")
    print("=" * 30)

split_bill(1850, 10, 4)
```

## App 4: Text Analyzer

```python
def analyze_text(text):
    words = text.split()
    letters = sum(ch.isalpha() for ch in text)
    digits = sum(ch.isdigit() for ch in text)
    spaces = text.count(" ")
    vowels = sum(ch in "aeiouAEIOU" for ch in text)
    print(f"Characters : {len(text)}")
    print(f"Words      : {len(words)}")
    print(f"Letters    : {letters}")
    print(f"Digits     : {digits}")
    print(f"Spaces     : {spaces}")
    print(f"Vowels     : {vowels}")
    print(f"Avg word length: {sum(len(w.strip('.,!?')) for w in words) / len(words):.2f}")
    print(f"Capitalized: {text.title()}")

analyze_text("Python 3 is fun! I wrote 100 lines of code today.")
```

## App 5: Password Strength Checker

```python
def check_password(pw):
    score = 0
    tips = []
    if len(pw) >= 8:
        score += 1
    else:
        tips.append("use at least 8 characters")
    if any(c.islower() for c in pw):
        score += 1
    else:
        tips.append("add lowercase letters")
    if any(c.isupper() for c in pw):
        score += 1
    else:
        tips.append("add uppercase letters")
    if any(c.isdigit() for c in pw):
        score += 1
    else:
        tips.append("add digits")
    if any(not c.isalnum() for c in pw):
        score += 1
    else:
        tips.append("add a symbol like @ # $")

    levels = {0: "Very weak", 1: "Very weak", 2: "Weak", 3: "Medium", 4: "Strong", 5: "Very strong"}
    return levels[score], tips

for pw in ["abc", "password1", "Python@2024"]:
    level, tips = check_password(pw)
    print(f"{pw!r:15} -> {level}", ("| tips: " + ", ".join(tips)) if tips else "")
```

## App 6: Student Report Card (types + operators + formatting)

```python
def report_card(name, marks):
    total = sum(marks.values())
    average = total / len(marks)
    passed = all(m >= 35 for m in marks.values())
    grade = "A" if average >= 80 else "B" if average >= 60 else "C" if average >= 40 else "F"

    print(f"Student: {name.title()}")
    print("-" * 28)
    for subject, m in marks.items():
        print(f"{subject.capitalize():<12}{m:>6}")
    print("-" * 28)
    print(f"{'Total':<12}{total:>6}")
    print(f"{'Average':<12}{average:>6.1f}")
    print(f"{'Grade':<12}{grade:>6}")
    print(f"{'Result':<12}{'PASS' if passed else 'FAIL':>6}")

report_card("asha kumar", {"maths": 88, "science": 76, "english": 69})
```
