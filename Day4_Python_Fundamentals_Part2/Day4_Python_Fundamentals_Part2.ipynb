# Part 1: Conditional Statements

## 1.1 The `if` statement

### 🧠 The idea
A program normally runs top to bottom. An `if` lets it **choose** whether to run some code.

> Real life: *"If it is raining, take an umbrella."*

### 📐 Syntax
```python
if condition:
    # indented code runs only when condition is True
```

⚠️ Two important things:
1. End the `if` line with a **colon** `:`
2. The code inside must be **indented** (press `Tab` or use 4 spaces)

````python
temperature = 35

if temperature > 30:
    print("It's hot! Drink water 💧")

print("This line always runs (it is not indented)")
````

**What happened?**
- Python checked `temperature > 30` → `True`
- So it ran the indented line
- The last `print` is not indented, so it runs no matter what

🔬 **Try it:** change `temperature` to `20` and run again. What changes?

## 1.2 `if` ... `else`

### 🧠 The idea
`else` is the "otherwise" plan: it runs when the condition is **False**.

> *"If it is raining, take an umbrella, otherwise wear sunglasses."*

````python
age = 16

if age >= 18:
    print("You can vote ✅")
else:
    print("You cannot vote yet ❌")
````

Exactly **one** of the two blocks runs, never both, never neither.

### Example: checking a password

````python
correct_password = "python123"
entered = "python123"

if entered == correct_password:
    print("Access granted 🔓")
else:
    print("Wrong password 🔒")
````

## 1.3 `if` ... `elif` ... `else`

### 🧠 The idea
When you have **more than two** choices, use `elif` (short for "else if").

> Traffic light: **if** red → stop, **elif** yellow → get ready, **else** → go.

Python checks from top to bottom and runs the **first** block whose condition is true. Then it skips the rest.

````python
marks = 72

if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
elif marks >= 40:
    grade = "C"
else:
    grade = "Fail"

print(f"Marks: {marks}  ->  Grade: {grade}")
````

**Trace it like a detective 🔍** (marks = 72):
1. `72 >= 90`? No → skip
2. `72 >= 75`? No → skip
3. `72 >= 60`? **Yes** → grade = "B", and Python ignores the rest

⚠️ **Order matters!** If you wrote `marks >= 40` first, everyone above 40 would get "C".

## 1.4 Combining conditions: `and`, `or`, `not`

| Word | Meaning | Example |
|------|---------|---------|
| `and` | **both** must be true | `age >= 18 and has_id` |
| `or` | **at least one** is true | `is_weekend or is_holiday` |
| `not` | flips True/False | `not is_raining` |

````python
age = 20
has_id = True

if age >= 18 and has_id:
    print("Entry allowed 🎟️")
else:
    print("Entry denied")

is_weekend = False
is_holiday = True

if is_weekend or is_holiday:
    print("No school today 🎉")

is_raining = False
if not is_raining:
    print("Let's go for a walk 🚶")
````

## 1.5 Nested `if` (an `if` inside an `if`)

# Part 2: `for` Loops

## 🧠 The idea
A **loop** repeats code so you don't have to write it again and again.

> Real life: *"For each student in the class, give a notebook."*

Without loops, printing 1 to 5 would need 5 print lines. With a loop it's 2 lines.

## 2.1 Looping through a list

````python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print("I like", fruit)
````

**How it works**
1. `fruit` takes the first item (`"apple"`), runs the indented block
2. Then it takes the next item (`"banana"`), runs the block again
3. Continues until the list is finished

`fruit` is just a variable name you choose. You could write `for item in fruits`.

## 2.2 Looping through a string
A string is a sequence of characters, so you can loop over it too.

````python
for letter in "Python":
    print(letter)
````

## 2.3 Using `range()` to repeat numbers

`range()` creates a sequence of numbers.

| Form | Meaning | Example result |
|------|---------|----------------|
| `range(5)` | 0 up to 4 | 0,1,2,3,4 |
| `range(1, 6)` | 1 up to 5 | 1,2,3,4,5 |
| `range(0, 10, 2)` | start, stop, step | 0,2,4,6,8 |
| `range(5, 0, -1)` | counting down | 5,4,3,2,1 |

The **stop** value is **not included**.

````python
print("range(5):")
for i in range(5):
    print(i, end=" ")

print("\n\nrange(1, 6):")
for i in range(1, 6):
    print(i, end=" ")

print("\n\nrange(0, 10, 2):")
for i in range(0, 10, 2):
    print(i, end=" ")

print("\n\nCountdown:")
for i in range(5, 0, -1):
    print(i, end=" ")
print("🚀 Lift off!")
````

## 2.4 Using a loop to calculate (accumulator pattern)

Very common idea: start with `total = 0`, then add to it inside the loop.

````python
total = 0
for number in range(1, 6):      # 1,2,3,4,5
    total = total + number
    print(f"Added {number}, total is now {total}")

print("Sum of 1 to 5 =", total)
````

# Part 3: `while` Loops

## 🧠 The idea
A `for` loop is used when you know **how many times** to repeat.
A `while` loop is used when you want to repeat **until something changes**.

> Real life: *"While the pot is not full, keep pouring water."*

### 📐 Syntax
```python
while condition:
    # runs again and again while condition is True
```

⚠️ Something inside the loop must eventually make the condition **False**, otherwise you get an **infinite loop**.

## 3.1 A simple counter

````python
count = 1

while count <= 5:
    print("Count is", count)
    count = count + 1       # without this line the loop never ends!

print("Loop finished")
````

**Step by step**
1. `count = 1` → is `1 <= 5`? Yes → print, then count becomes 2
2. `2 <= 5`? Yes → print, count becomes 3
3. ... continues until count becomes 6
4. `6 <= 5`? No → loop stops

## 3.2 Countdown

````python
n = 5
while n > 0:
    print(n)
    n -= 1                  # same as n = n - 1
print("Go! 🏁")
````

## 3.3 Repeat until the user gives valid input

This is where `while` shines. We keep asking until we get a good answer.
(In non-interactive mode we simulate answers from a list.)

````python
# Simulated answers so this cell runs by itself
sample_answers = iter(["abc", "-5", "25"])

def get_age():
    while True:
        text = input("Enter your age: ") if RUN_INTERACTIVE else next(sample_answers)
        if not RUN_INTERACTIVE:
            print("Enter your age:", text, "  (sample)")

        if not text.isdigit():
            print("❌ Please type digits only.")
        elif int(text) < 1 or int(text) > 120:
            print("❌ Age must be 1 to 120.")
        else:
            return int(text)

age = get_age()
print("Your age is", age)
````

## 3.4 `for` or `while`? How to choose

| Situation | Use |
|-----------|-----|
| Loop through a list / string / range | `for` |
| Known number of repeats | `for` |
| Repeat until a condition changes | `while` |
| Unknown number of repeats (e.g. user input) | `while` |

## 3.5 Sum of digits (a classic `while` example)

````python
number = 4582
total = 0

while number > 0:
    last_digit = number % 10      # get last digit
    total += last_digit
    number = number // 10         # remove last digit
    print(f"digit={last_digit}, remaining number={number}, total={total}")

print("Sum of digits =", total)
````

# Part 4: `break` and `continue`

These two keywords give you extra control inside loops.

| Keyword | What it does | Real-life |
|---------|--------------|-----------|
| `break` | Stops the **whole loop** immediately | "Found my keys, stop searching!" |
| `continue` | Skips **this round** and moves to the next | "Skip the rotten apple, check the next one" |

## 4.1 `break`

````python
for number in range(1, 11):
    if number == 6:
        print("Found 6, stopping!")
        break
    print(number)
print("Loop ended")
````

````python
# Searching a list
names = ["Asha", "Ravi", "Meena", "John"]
search = "Meena"

for name in names:
    if name == search:
        print(f"{search} found! 🎉")
        break
else:
    # this else belongs to the FOR loop: runs only if break did NOT happen
    print(f"{search} not found")
````

💡 **Bonus:** a loop can have an `else`. It runs only when the loop finished normally (no `break`). It's handy for searching and for the prime checker later.

## 4.2 `continue`

````python
for number in range(1, 8):
    if number % 2 == 0:
        continue            # skip even numbers
    print(number)
````

````python
# Skip a specific word
for word in ["python", "java", "c++", "ruby"]:
    if word == "java":
        continue
    print(word)
````

## 4.3 Using `break` with `while True`
`while True:` loops forever, so we use `break` to exit when ready. This is a very common pattern for menus and input.

````python
commands = iter(["hello", "how are you", "quit", "this is never reached"])

while True:
    cmd = input("Type something (quit to exit): ") if RUN_INTERACTIVE else next(commands)
    if not RUN_INTERACTIVE:
        print("Type something (quit to exit):", cmd, "  (sample)")

    if cmd == "quit":
        print("Bye 👋")
        break
    print("You typed:", cmd)
````

### ✅ Mini-check: `break` vs `continue`
Run it and notice the difference.

````python
print("With break:")
for i in range(1, 6):
    if i == 3:
        break
    print(i, end=" ")

print("\nWith continue:")
for i in range(1, 6):
    if i == 3:
        continue
    print(i, end=" ")
````

# Part 5: Nested Loops

## 🧠 The idea
A loop **inside** another loop.

> Clock: for each **hour**, the **minutes** go through 0 to 59. The inner loop completes fully for every single step of the outer loop.

## 5.1 Seeing how it runs

````python
for i in range(1, 4):           # outer loop: 3 times
    for j in range(1, 3):       # inner loop: 2 times for every outer turn
        print(f"i={i}, j={j}")
    print("--- inner loop finished ---")
````

Total prints = 3 × 2 = 6. The inner loop restarts from the beginning each time the outer loop moves.

## 5.2 Multiplication table grid

````python
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i * j:>3}", end=" ")
    print()      # new line after each row
````

## 5.3 Star patterns (great practice!)

**Right triangle**

````python
rows = 5
for i in range(1, rows + 1):
    for j in range(i):
        print("*", end=" ")
    print()
````

**Number pyramid (using nested loops)**

````python
rows = 5
for i in range(1, rows + 1):
    print(" " * (rows - i), end="")
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
````

**Square of stars**

````python
size = 4
for row in range(size):
    for col in range(size):
        print("#", end=" ")
    print()
````

## 5.4 `break` inside nested loops
`break` only exits the **innermost** loop it is in.

````python
for i in range(1, 4):
    for j in range(1, 4):
        if j == 2:
            break           # leaves only the inner loop
        print(f"i={i}, j={j}")
````

# Part 6: List Comprehension

## 🧠 The idea
A short, neat way to **build a new list** from a loop in a single line.

### First, the long way

````python
squares = []
for n in range(1, 6):
    squares.append(n ** 2)
print(squares)
````

### Now the short way
```python
[expression  for item in sequence]
```
Read it like English: *"give me n squared, for each n in 1 to 5."*

````python
squares = [n ** 2 for n in range(1, 6)]
print(squares)
````

## 6.1 More examples

````python
# Double every number
numbers = [1, 2, 3, 4, 5]
doubled = [n * 2 for n in numbers]
print(doubled)

# Uppercase every name
names = ["asha", "ravi", "meena"]
print([name.upper() for name in names])

# Length of each word
words = ["python", "is", "fun"]
print([len(w) for w in words])

# Characters of a string
print([ch for ch in "hello"])
````

## 6.2 Adding a condition (filter)
```python
[expression  for item in sequence  if condition]
```
*"...but only keep the items where the condition is true."*

````python
numbers = list(range(1, 21))

evens = [n for n in numbers if n % 2 == 0]
odds = [n for n in numbers if n % 2 != 0]
big = [n for n in numbers if n > 15]

print("Evens:", evens)
print("Odds :", odds)
print("Big  :", big)
````

````python
# Words longer than 3 letters
words = ["a", "cat", "python", "is", "awesome"]
print([w for w in words if len(w) > 3])

# Marks that passed
marks = [35, 72, 48, 90, 20]
print([m for m in marks if m >= 40])
````

## 6.3 `if ... else` inside a comprehension
When you want to **change** the value (not filter), put `if else` **before** the `for`.

````python
numbers = [1, 2, 3, 4, 5, 6]
labels = ["even" if n % 2 == 0 else "odd" for n in numbers]
print(labels)

marks = [35, 72, 48, 90]
print(["Pass" if m >= 40 else "Fail" for m in marks])
````

## 6.4 Nested comprehension (a small grid)

````python
table = [[i * j for j in range(1, 4)] for i in range(1, 4)]
for row in table:
    print(row)
````

💡 **Tip for beginners:** if a comprehension feels confusing, write it as a normal loop first, then convert it. Readability is more important than being short.

# 🛠️ Practical 1: Even and Odd Number Identification

### 🧠 Logic
A number is **even** if dividing by 2 leaves **remainder 0**, otherwise it is odd.
We use the **modulus operator** `%` to get the remainder.

| Number | `number % 2` | Result |
|--------|--------------|--------|
| 8 | 0 | Even |
| 7 | 1 | Odd |

## Step 1: Check one number

````python
number = int(ask("Enter a number: ", "17"))

if number % 2 == 0:
    print(f"{number} is Even")
else:
    print(f"{number} is Odd")
````

## Step 2: Turn it into a reusable function

````python
def even_or_odd(n):
    if n % 2 == 0:
        return "Even"
    return "Odd"

for value in [0, 1, 10, 33, -4]:
    print(f"{value:>3} -> {even_or_odd(value)}")
````

## Step 3: Check every number in a range using a loop

````python
for n in range(1, 11):
    print(f"{n:>2} is {even_or_odd(n)}")
````

## Step 4: Separate evens and odds from a list

````python
numbers = [12, 7, 45, 30, 8, 91, 64, 3]

evens = []
odds = []
for n in numbers:
    if n % 2 == 0:
        evens.append(n)
    else:
        odds.append(n)

print("Evens:", evens)
print("Odds :", odds)

# Same thing with list comprehension
print("Evens:", [n for n in numbers if n % 2 == 0])
print("Odds :", [n for n in numbers if n % 2 != 0])
print("Sum of evens:", sum(n for n in numbers if n % 2 == 0))
````

# 🛠️ Practical 2: Prime Number Checker

### 🧠 What is a prime number?
A number greater than 1 that is divisible **only by 1 and itself**.
Examples: 2, 3, 5, 7, 11, 13. Non-examples: 4 (2×2), 9 (3×3), 1 (not prime by definition).

### 🧠 Plan (in plain English)
1. If the number is less than 2 → not prime
2. Try dividing by 2, 3, 4, ... up to the number − 1
3. If any number divides it **exactly** (remainder 0) → not prime, stop
4. If none divide it → prime

## Step 1: Simple version

````python
number = int(ask("Enter a number: ", "29"))

if number < 2:
    print(f"{number} is not prime")
else:
    is_prime = True                  # assume prime until proven otherwise
    for i in range(2, number):
        if number % i == 0:          # found a divisor
            is_prime = False
            break                    # no need to check more
    if is_prime:
        print(f"{number} is a prime number ✅")
    else:
        print(f"{number} is NOT a prime number ❌ (divisible by {i})")
````

## Step 2: Make it a function and test many numbers

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

for n in [0, 1, 2, 3, 4, 17, 25, 29, 97, 100]:
    print(f"{n:>3} -> {'Prime' if is_prime(n) else 'Not prime'}")
````

## Step 3: Faster version
We only need to check up to the **square root** of the number. If a number has a factor larger than its square root, it must also have a smaller one.

````python
def is_prime_fast(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:        # same as i <= sqrt(n)
        if n % i == 0:
            return False
        i += 1
    return True

print(is_prime_fast(97), is_prime_fast(7919), is_prime_fast(1000001))
````

## Step 4: Find all primes in a range (loop + comprehension)

````python
primes = [n for n in range(1, 51) if is_prime_fast(n)]
print("Primes up to 50:", primes)
print("How many:", len(primes))
````

## Step 5: Show factors using nested loops

````python
for n in range(2, 16):
    factors = [i for i in range(1, n + 1) if n % i == 0]
    kind = "prime" if len(factors) == 2 else "composite"
    print(f"{n:>2}: factors={factors}  -> {kind}")
````

# 🛠️ Practical 3: Number Guessing Game 🎯

### 🧠 How the game works
1. The computer secretly picks a number between 1 and 100
2. You guess
3. The computer says **"Too low"** or **"Too high"**
4. Continue until you guess it or run out of attempts

### Concepts used
`random`, `while` loop, `if / elif / else`, `break`, counters.

## Step 1: Pick a random number

````python
secret = random.randint(1, 100)       # a different number each time
print("The computer picked a number (hidden).")
# print(secret)  # remove the # to cheat while testing
````

## Step 2: One guess only (decision making)

````python
secret = 42
guess = 30

if guess == secret:
    print("Correct! 🎉")
elif guess < secret:
    print("Too low 📉")
else:
    print("Too high 📈")
````

## Step 3: Full game with a loop

We put the game in a **function** so we can run it either with typed guesses (interactive) or with a list of sample guesses (so the notebook can demo it).

````python
def guessing_game(secret, max_attempts=7, sample_guesses=None):
    print("🎯 Guess the number between 1 and 100")
    print(f"You have {max_attempts} attempts.\n")

    attempts = 0
    guesses = iter(sample_guesses) if sample_guesses else None

    while attempts < max_attempts:
        # get a guess
        if guesses is None:
            text = input(f"Attempt {attempts + 1}: ")
        else:
            text = str(next(guesses))
            print(f"Attempt {attempts + 1}: {text}   (sample)")

        # validate
        if not text.isdigit():
            print("❌ Enter a whole number (this does not count as an attempt)\n")
            continue

        guess = int(text)
        attempts += 1

        # compare
        if guess == secret:
            print(f"🎉 Correct! You got it in {attempts} attempts.")
            return True
        elif guess < secret:
            print("📉 Too low\n")
        else:
            print("📈 Too high\n")

    print(f"😢 Out of attempts. The number was {secret}.")
    return False

# Demo with fixed guesses (a smart player uses halving: 50, 25, 37 ...)
guessing_game(secret=37, sample_guesses=[50, 25, "abc", 37])
````

````python
# Play it yourself! (set RUN_INTERACTIVE = True in the first cell)
if RUN_INTERACTIVE:
    guessing_game(secret=random.randint(1, 100))
else:
    print("Set RUN_INTERACTIVE = True at the top and re-run this cell to play.")
````

## Step 4: Game demo that plays itself (binary search idea)
Smart strategy: always guess the **middle** of the remaining range. You can find any number in at most 7 guesses between 1 and 100.

````python
secret = random.randint(1, 100)
low, high = 1, 100
attempts = 0

while True:
    guess = (low + high) // 2
    attempts += 1
    print(f"Guess {attempts}: {guess}", end=" -> ")
    if guess == secret:
        print("Correct! 🎉")
        break
    elif guess < secret:
        print("Too low")
        low = guess + 1
    else:
        print("Too high")
        high = guess - 1

print(f"Secret was {secret}, found in {attempts} guesses")
````

## App 1: Grade Calculator (if / elif / else + loop)

````python
def get_grade(mark):
    if mark >= 90: return "A+"
    elif mark >= 80: return "A"
    elif mark >= 70: return "B"
    elif mark >= 50: return "C"
    elif mark >= 35: return "D"
    else: return "F"

students = {"Asha": 92, "Ravi": 78, "Meena": 55, "John": 30}

print(f"{'Name':<8}{'Mark':>5}  Grade")
print("-" * 22)
for name, mark in students.items():
    print(f"{name:<8}{mark:>5}  {get_grade(mark)}")

print("\nToppers (80+):", [n for n, m in students.items() if m >= 80])
print("Average mark :", sum(students.values()) / len(students))
````

## App 2: FizzBuzz (the famous interview warm-up)

````python
# Multiples of 3 -> Fizz, multiples of 5 -> Buzz, both -> FizzBuzz
for n in range(1, 21):
    if n % 15 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
````

## App 3: Simple ATM (while loop + menu)

````python
def atm(actions):
    balance = 1000
    actions = iter(actions)
    while True:
        print("\n1) Check balance  2) Deposit  3) Withdraw  4) Exit")
        choice = input("Choose: ") if RUN_INTERACTIVE else next(actions)
        if not RUN_INTERACTIVE:
            print("Choose:", choice, "  (sample)")

        if choice == "1":
            print("Balance:", balance)
        elif choice == "2":
            amount = int(input("Amount: ") if RUN_INTERACTIVE else next(actions))
            balance += amount
            print("Deposited. New balance:", balance)
        elif choice == "3":
            amount = int(input("Amount: ") if RUN_INTERACTIVE else next(actions))
            if amount > balance:
                print("❌ Insufficient funds")
            else:
                balance -= amount
                print("Withdrawn. New balance:", balance)
        elif choice == "4":
            print("Thank you! 👋")
            break
        else:
            print("Invalid choice")

# Sample run: check, deposit 500, withdraw 2000 (fails), withdraw 300, exit
atm(["1", "2", "500", "3", "2000", "3", "300", "4"]) if not RUN_INTERACTIVE else atm([])
````

## App 4: Login with limited attempts (while + break)

````python
def login(attempts_list, correct="python123", max_tries=3):
    tries = 0
    entries = iter(attempts_list)
    while tries < max_tries:
        pw = input("Password: ") if RUN_INTERACTIVE else next(entries)
        if not RUN_INTERACTIVE:
            print("Password:", pw, "  (sample)")
        tries += 1
        if pw == correct:
            print("✅ Login successful")
            return
        print(f"❌ Wrong password ({max_tries - tries} tries left)")
    print("🔒 Account locked")

login(["abc", "123", "python123"])
print()
login(["a", "b", "c"])
````

## App 5: Word Filter (list comprehension)

````python
sentence = "Python is a very powerful and simple programming language"
words = sentence.split()

print("Long words (>5)  :", [w for w in words if len(w) > 5])
print("Capitalised      :", [w.capitalize() for w in words])
print("Words with 'o'   :", [w for w in words if "o" in w])
print("First letters    :", "".join(w[0] for w in words))
````

## App 6: Multiplication Tables 1 to 5 (nested loops)

````python
print("    " + "".join(f"{i:>4}" for i in range(1, 11)))
print("    " + "-" * 40)
for t in range(1, 6):
    print(f"{t:>2} |" + "".join(f"{t * i:>4}" for i in range(1, 11)))
````

# 📌 Summary and Cheat Sheet

```python
# Decisions
if cond1:
    ...
elif cond2:
    ...
else:
    ...

# Loops
for item in sequence:        # known sequence
    ...
for i in range(1, 6):        # 1 to 5
    ...
while condition:             # repeat until False
    ...

# Control
break                        # exit loop
continue                     # skip to next round

# Nested
for i in range(3):
    for j in range(3):
        ...

# List comprehension
[expr for x in seq]
[expr for x in seq if cond]
[a if cond else b for x in seq]
```

| Concept | Key point |
|---------|-----------|
| `if / elif / else` | Python runs the first true branch only |
| `for` | Use when you know what to loop through |
| `while` | Use when you repeat until something changes (update the variable!) |
| `break` | Stop the loop completely |
| `continue` | Skip the rest of this round |
| Nested loops | Inner loop finishes fully for each outer step |
| List comprehension | Short way to build or filter lists |
| `%` | Remainder: powers even/odd and prime checks |
